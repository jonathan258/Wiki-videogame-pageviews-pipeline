```sql
SELECT
    check_run_id,
    check_name,
    check_status,
    expected_value,
    actual_value,
    details,
    checked_at
FROM wiki_videogame_ingestion.bronze.dq_check_results
WHERE check_run_id = (
    SELECT MAX(check_run_id)
    FROM wiki_videogame_ingestion.bronze.dq_check_results
)
ORDER BY check_name;
```
